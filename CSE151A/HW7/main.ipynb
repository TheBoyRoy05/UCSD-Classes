```python
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

# Question 7

```python
def classify(x, w, b):
    return np.sign(np.dot(x, w) + b)

def perceptron(X, Y, epochs=100):
    perm = np.random.permutation(len(X))
    X = X[perm]
    Y = Y[perm]

    w = np.zeros(X.shape[1])
    b = 0
    updates_per_epoch = []

    for _ in range(epochs):
        updates = 0
        for x, y in zip(X, Y):
            if classify(x, w, b) != y:
                w += y * x
                b += y
                updates += 1
        updates_per_epoch.append(updates)
        if updates == 0:
            break

    return w, b, updates_per_epoch

def plot_perceptron(X, Y, w, b):
    fig = px.scatter(x=X[:, 0], y=X[:, 1], color=Y.astype(str),
                 labels={'x': 'Sepal Width', 'y': 'Petal Width'},
                 title='Iris Data with Perceptron Decision Boundary')

    x_range = np.linspace(X[:, 0].min(), X[:, 0].max(), 100)
    y_range = -(w[0] * x_range + b) / w[1]

    fig.add_trace(go.Scatter(x=x_range, y=y_range, mode='lines', 
                            name='Decision Boundary', line=dict(color='black')))

    fig.show()
```

```python
from sklearn import datasets
iris = datasets.load_iris()
X = iris.data
Y = iris.target

X = X[Y != 2]
X = X[:, [1, 3]]
Y = Y[Y != 2]
Y[Y == 0] = -1
```

```python
w, b, _ = perceptron(X, Y)
plot_perceptron(X, Y, w, b)
```

```python
updates = []
for i in range(20):
    _, _, update = perceptron(X, Y)
    updates.append(sum(update))

fig = px.histogram(updates, nbins=max(updates), title='Histogram of Number of Updates')
fig.show()
```

```python
from sklearn import datasets
iris = datasets.load_iris()
X = iris.data
Y = iris.target

X = X[Y != 0]
X = X[:, [0, 2]]
Y = Y[Y != 0]
Y[Y == 2] = -1
```

```python
w, b, _ = perceptron(X, Y)
plot_perceptron(X, Y, w, b)
```

```python
from sklearn.svm import SVC

C_values = [0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000, 10000, 100000]
training_errors = []
support_vectors = []

for C in C_values:
    svm = SVC(kernel='linear', C=C)
    svm.fit(X, Y)
    training_errors.append(round(1 - svm.score(X, Y), 2))
    support_vectors.append(int(svm.n_support_.sum()))

training_errors, support_vectors
```
