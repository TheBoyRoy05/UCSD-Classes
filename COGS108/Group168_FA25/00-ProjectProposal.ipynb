## Research Question

Our research investigates how effectively Body Mass Index (BMI) predicts overall health outcomes compared to lifestyle factors such as daily calorie intake, exercise frequency, and hours of sleep. Specifically, we aim to determine whether these behavioral indicators are stronger predictors of health status—measured by the presence of diabetes, high blood pressure, or high cholesterol—than BMI alone. Prior studies suggest that while BMI is widely used as a health indicator, it may not fully capture the effects of lifestyle behaviors on metabolic health. To address this, we will analyze the relationships between BMI, lifestyle factors, and health outcomes using correlation and multiple regression analyses, controlling for demographic variables such as age and sex. This project will evaluate the predictive validity of BMI relative to lifestyle-based health metrics to determine whether BMI is a sufficient standalone indicator of overall health.

## Background and Prior Work

For years, the concept of BMI, or Body Mass Index, has been used as a reliable and quick way to determine an individual’s general health status. This numerical value, derived from a person’s weight and height, provides a simple estimate of whether someone is underweight, at a healthy weight, overweight, or obese. However, while BMI is helpful for establishing broad population trends, it fails to account for key individual differences such as fat distribution, metabolic rate, and muscle mass, which can significantly influence overall health outcomes.

In the past, multiple studies have explored what physical and biochemical factors most accurately predict chronic health conditions. One such study, “Predicting the Onset of Diabetes with Machine Learning Methods,” published in Frontiers in Public Health, used advanced machine learning models to identify risk factors contributing to diabetes onset. The best performing models, two-class decision jungle and two-class boosted decision tree, achieved AUC scores of 0.976 and 0.991, demonstrating highly accurate prediction capabilities based on diverse health metrics rather than BMI alone.<sup><a href="#ref1">1</a></sup>

Similarly, a Nature study titled “Identifying Top Ten Predictors of Type 2 Diabetes through Machine Learning Analysis of UK Biobank Data” (2024) used large scale data and predictive modeling to determine which biological and lifestyle variables most strongly indicate diabetes risk. The top predictors for males were HbA1c, plasma glucose, BMI, family history of diabetes, and GGT, suggesting that biochemical markers outperform BMI in predicting disease outcomes.<sup><a href="#ref2">2</a></sup>

Beyond published studies, open-source data projects have also investigated diabetes prediction. A Kaggle project titled “Diabetes Prediction” by Abdollahii tested multiple machine learning models such as CatBoost, SVM, XGBoost, and Random Forest on clinical and behavioral data. The CatBoost model achieved the highest AUC, with all others scoring above 0.95, indicating that ensemble models effectively capture the multifaceted nature of diabetes risk, including BMI and metabolic factors.<sup><a href="#ref3">3</a></sup>

Another Kaggle analysis, “Diabetes EDA & Random Forest HP” by Tumpan Jawat, focused on feature importance analysis using a Random Forest classifier. The study found HbA1c level to be the most important predictor of diabetes (importance = 0.44), followed by blood glucose level (0.32), age (0.14), and BMI (0.06). Other factors, including hypertension and heart disease, showed minor predictive relevance, while smoking history and gender had minimal impact. These results reinforce that although BMI plays a role in diabetes prediction, biochemical and age-related variables are far stronger indicators of disease likelihood.<sup><a href="#ref4">4</a></sup>

## Hypothesis


We predict that BMI alone is a weaker predictor of overall health outcomes than behavioral and biochemical indicators such as calorie intake, hours of sleep, exercise frequency, and blood glucose levels. We expect this because prior research suggests that BMI does not distinguish between muscle and fat mass or account for behavioral habits that directly affect metabolic health. In contrast, lifestyle habits directly influence energy balance, cardiovascular function, and metabolic regulation, making them more accurate indicators of overall health. Therefore, individuals with healthier lifestyles may exhibit better health outcomes even if their BMI falls outside the “normal” range.

## Data

### Ideal Data

To answer our research question, the ideal dataset would include a large and diverse sample of individuals with detailed health, lifestyle, and demographic information. Key variables would include Body Mass Index (BMI), daily calorie intake, exercise frequency, hours of sleep, blood pressure, cholesterol levels, blood glucose, presence of diabetes, high blood pressure, cardiovascular disease, age, sex, and other relevant demographic or lifestyle factors. The dataset should have at least 5,000-10,000 observations to ensure statistical power for multiple regression analyses and to allow generalization across populations.

### Potential real datasets include:

Diabetes Dataset - Kaggle
URL: https://www.kaggle.com/datasets/akshaydattatraykhare/diabetes-dataset

This dataset contains 768 observations and includes several variables relevant to health outcomes. It records the number of pregnancies, blood glucose level, blood pressure measurement, skin thickness, blood insulin level, Body Mass Index (BMI), family history-based diabetes risk percentage (Diabetes Pedigree Function), age, and a binary indicator for diabetes outcome, where 1 represents a diagnosis of diabetes and 0 represents no diagnosis. This dataset provides both clinical indicators, such as BMI, glucose, and insulin, and basic demographic information, making it suitable for evaluating the predictive power of BMI relative to other health factors. It can be downloaded directly after creating a free Kaggle account.

Diabetes Health Indicators Dataset - Kaggle
URL: https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset/data

This dataset contains hundreds of thousands of observations and includes a wide range of health, lifestyle, and demographic variables. Key variables include a binary diabetes indicator, high blood pressure, high cholesterol, cholesterol check, BMI, smoking status, history of stroke, heart disease or attack, physical activity, fruit and vegetable consumption, heavy alcohol consumption, healthcare access, inability to see a doctor due to cost, general health, mental health, physical health, difficulty walking, sex, age, education level, and income. This dataset is particularly useful for analyzing how lifestyle and demographic factors, alongside BMI, influence health outcomes. The dataset can be downloaded after creating a free Kaggle account.

Diabetes Prediction Dataset - Kaggle
URL: https://www.kaggle.com/datasets/iammustafatz/diabetes-prediction-dataset

This dataset contains 100,000 observations and includes features such as age, gender, Body Mass Index (BMI), hypertension, heart disease, smoking history, HbA1c level, and blood glucose level. The dataset provides both clinical indicators and lifestyle factors, making it suitable for evaluating the predictive power of BMI compared to other behavioral and biochemical health indicators. It can be accessed and downloaded directly after registering on Kaggle.
