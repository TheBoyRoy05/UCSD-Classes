```python
import numpy as np
import plotly.subplots as sp
import plotly.graph_objects as go
```

```python
def plot(losses, constraints, title=None):
    fig = sp.make_subplots(rows=1, cols=2)

    fig.add_trace(
        go.Scatter(
            x=np.arange(len(losses)), 
            y=losses, 
            mode="lines", 
            name="Loss"
        ), 
        row=1, 
        col=1
    )

    fig.add_trace(
        go.Scatter(
            x=np.arange(len(constraints)), 
            y=constraints, 
            mode="lines", 
            name="Constraint Violation"
        ),
        row=1,
        col=2,
    )

    fig.update_layout(
        showlegend=True,
        height=400,
        width=1000,
        title_text=f"Loss and Constraint Violation vs. Iteration {title and f'({title})'}",
        title_x=0.5,
        xaxis_title_text="Iteration",
        xaxis2_title_text="Iteration",
        yaxis_title_text="Loss",
        yaxis2_title_text="Constraint Violation",
    )

    fig.update_yaxes(type="log", row=1, col=1)
    fig.update_yaxes(type="log", row=1, col=2)

    fig.show()
```

```python
# Question 2c
n = 1000
m = 500
A = np.random.normal(size=(m, n))
b = np.ones(m)
alpha = 3.8 / np.linalg.norm(A, 2) ** 2
T = 600
w = np.zeros(m)

constraints = np.array([])
losses = np.array([])

for t in range(T):
    x = A.T @ w / 2
    w = w + alpha * (-A @ x + b)
    losses = np.append(losses, np.dot(x, x))
    constraints = np.append(constraints, np.linalg.norm(A @ x - b))
    
plot(losses, constraints, title="alpha = 3.8 / ||A||^2")
```

```python
# Question 2d
n = 1000
m = 500
A = np.random.normal(size=(m, n))
b = np.ones(m)
rho = 1e3
T = 3
w = np.zeros(m)

constraints = np.array([])
losses = np.array([])

for t in range(T):
    x = np.linalg.inv(2 * np.eye(n) + rho * A.T @ A) @ (A.T @ w + rho * A.T @ b)
    w = w + rho * (-A @ x + b)
    losses = np.append(losses, np.dot(x, x))
    constraints = np.append(constraints, np.linalg.norm(A @ x - b))
    
plot(losses, constraints, title="rho = 1000")
```
